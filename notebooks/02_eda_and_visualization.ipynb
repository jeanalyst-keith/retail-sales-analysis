# 02 - Exploratory Data Analysis & Visualization
**Retail Sales Analysis**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots

import warnings
warnings.filterwarnings('ignore')

%matplotlib inline
plt.style.use('seaborn-v0_8')
```

```python
# Load the cleaned data
df = pd.read_csv('data/processed/cleaned_sales.csv')
print(df.shape)
df.head()
```

## 1. Overview of the Dataset

```python
print(df.describe())
print('\nUnique Product Categories:', df['Product Category'].unique())
```

## 2. Revenue by Product Category

```python
category_revenue = df.groupby('Product Category')['Revenue'].sum().reset_index()
fig = px.bar(category_revenue, x='Product Category', y='Revenue', 
             title='Total Revenue by Product Category',
             color='Product Category')
fig.show()
```

## 3. Average Spending by Gender

```python
gender_spending = df.groupby('Gender')['Revenue'].mean().reset_index()
fig = px.bar(gender_spending, x='Gender', y='Revenue', 
             title='Average Revenue per Transaction by Gender',
             color='Gender')
fig.show()
```

## 4. Sales Trend Over Time

```python
daily_sales = df.groupby('Date')['Revenue'].sum().reset_index()
fig = px.line(daily_sales, x='Date', y='Revenue', title='Daily Sales Trend')
fig.show()
```

## 5. Revenue by Age Group

```python
age_revenue = df.groupby('Age_Group')['Revenue'].sum().reset_index()
fig = px.bar(age_revenue, x='Age_Group', y='Revenue', 
             title='Total Revenue by Age Group', color='Age_Group')
fig.show()
```
