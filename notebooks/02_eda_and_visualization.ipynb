# 02 - Exploratory Data Analysis & Visualization
**Retail Sales Analysis Project**

```python
import pandas as pd
import os
import plotly.express as px

# Fix working directory
if os.getcwd().endswith('notebooks'):
    os.chdir('..')

print("Current working directory:", os.getcwd())

# Load cleaned data
df = pd.read_csv('data/processed/cleaned_sales.csv')
print("✅ Loaded cleaned dataset with shape:", df.shape)
```

## 1. Revenue by Product Category

```python
cat_revenue = df.groupby('Product Category')['Revenue'].sum().reset_index().sort_values('Revenue', ascending=False)

fig1 = px.bar(cat_revenue, x='Product Category', y='Revenue', 
              title='Total Revenue by Product Category',
              color='Product Category', text_auto=True)
fig1.update_layout(height=500)
fig1.show()
```

## 2. Average Spending by Gender

```python
gender_spend = df.groupby('Gender')['Revenue'].mean().reset_index()

fig2 = px.bar(gender_spend, x='Gender', y='Revenue', 
              title='Average Revenue per Transaction by Gender',
              color='Gender', text_auto=True)
fig2.show()
```

## 3. Revenue by Age Group

```python
age_revenue = df.groupby('Age_Group')['Revenue'].sum().reset_index()

fig3 = px.bar(age_revenue, x='Age_Group', y='Revenue', 
              title='Total Revenue by Age Group',
              color='Age_Group', text_auto=True)
fig3.show()
```
